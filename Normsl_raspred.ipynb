```python
import math
from scipy.stats import norm

# Параметры нормального распределения
mu = 5.6
sigma = math.sqrt(2.25)

# Границы интервала
a = 2.9
b = 8.3

# Вычисляем вероятность
probability = norm.cdf((b - mu) / sigma) - norm.cdf((a - mu) / sigma)

# Округляем результат до 4 знаков после запятой
print(f'P(2.9 <= ξ <= 8.3) = {round(probability, 4)}')
```
